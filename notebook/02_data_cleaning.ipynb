```python
import pandas as pd
import numpy as np
#导入健康数据汇总csv
df=pd.read_csv('D:/DA/Huawei_Health_Analysis/DATA/processed/health_daily_summary.csv')
#预览数据前五行
df.head()
```

```python
df.isnull().sum()
```

```python
#用各列的中位数填充缺失值
df['REM时长']=df['REM时长'].fillna(df['REM时长'].median())
df['REM时长']=df['REM时长'].round(1)

df['午睡时长']=df['午睡时长'].fillna(df['午睡时长'].median())
df['午睡时长']=df['午睡时长'].round(1)

df['深睡时长']=df['深睡时长'].fillna(df['深睡时长'].median())
df['深睡时长']=df['深睡时长'].round(1)

df['浅睡时长']=df['浅睡时长'].fillna(df['浅睡时长'].median())
df['浅睡时长']=df['浅睡时长'].round(1)

df['睡眠效率']=df['睡眠效率'].fillna(df['睡眠效率'].median())
df['睡眠效率']=df['睡眠效率'].round(2)

df['清醒时长']=df['清醒时长'].fillna(df['清醒时长'].median())
df['清醒时长']=df['清醒时长'].round(1)

df['夜间总睡眠']=df['夜间总睡眠'].fillna(df['夜间总睡眠'].median())
df['夜间总睡眠']=df['夜间总睡眠'].round(2)

df['深睡比例']=df['深睡比例'].fillna(df['深睡比例'].median())
df['深睡比例']=df['深睡比例'].round(2)

df['浅睡比例']=df['浅睡比例'].fillna(df['浅睡比例'].median())
df['浅睡比例']=df['浅睡比例'].round(2)

df['快速眼动比例']=df['快速眼动比例'].fillna(df['快速眼动比例'].median())
df['快速眼动比例']=df['快速眼动比例'].round(2)

df['在床时间']=df['在床时间'].fillna(df['在床时间'].median())
df['在床时间']=df['在床时间'].round(1)

df['总时长']=df['总时长'].fillna(df['总时长'].median())
df['总时长']=df['总时长'].round(1)

df['总步数']=df['总步数'].fillna(df['总步数'].median())
df['总步数']=df['总步数'].round(1)

df['基础代谢']=df['基础代谢'].fillna(df['基础代谢'].median())
df['基础代谢']=df['基础代谢'].round(1)

df['平均心率']=df['平均心率'].fillna(df['平均心率'].median())
df['平均心率']=df['平均心率'].round(2)

df['最大心率']=df['最大心率'].fillna(df['最大心率'].median())
df['最大心率']=df['最大心率'].round(1)

df['最小心率']=df['最小心率'].fillna(df['最小心率'].median())
df['最小心率']=df['最小心率'].round(1)

df['记录条数']=df['记录条数'].fillna(df['记录条数'].median())
df['记录条数']=df['记录条数'].round(1)

#用该列第一个众数填充缺失值
df['是否午睡']=df['是否午睡'].fillna(df['是否午睡'].mode()[0])

df['星期']=df['星期'].fillna(df['星期'].mode()[0])

df.isnull().sum()
```

```python
#查询重复列
dup=df.duplicated().sum()
print(dup)
```

```python
#查看各字段基础统计值，查看极端值是否合理
df.describe()
```

```python
#处理异常值
df['REM时长']=df['REM时长'].replace(0,np.nan)

df['浅睡时长']=df['浅睡时长'].replace(0,np.nan)

df['深睡时长']=df['深睡时长'].replace(0,np.nan)

df['夜间总睡眠']=df['夜间总睡眠'].replace(0,np.nan)

df['在床时间']=df['在床时间'].replace(0,np.nan)

df['活动消耗']=df['活动消耗'].replace(0,np.nan)

df['全天总消耗']=df['全天总消耗'].replace(0,np.nan)
#再次查看基础统计值
df.describe()
```

```python
#活动消耗和全天总消耗最小值异常，经查2026-03-24当天导出数据时未记录完全整体消耗，删除该日。
df=df.drop(df[df['日期']=='2026-03-24'].index)
```

```python
#再次查看缺失值情况
df.isnull().sum()
```

```python
#再次填充空缺值
df['REM时长']=df['REM时长'].fillna(df['REM时长'].median())
df['REM时长']=df['REM时长'].round(1)

df['深睡时长']=df['深睡时长'].fillna(df['深睡时长'].median())
df['深睡时长']=df['深睡时长'].round(1)

df['浅睡时长']=df['浅睡时长'].fillna(df['浅睡时长'].median())
df['浅睡时长']=df['浅睡时长'].round(1)

df['夜间总睡眠']=df['夜间总睡眠'].fillna(df['夜间总睡眠'].median())
df['夜间总睡眠']=df['夜间总睡眠'].round(2)

df['在床时间']=df['在床时间'].fillna(df['在床时间'].median())
df['在床时间']=df['在床时间'].round(1)

df['活动消耗']=df['活动消耗'].fillna(df['活动消耗'].median())
df['活动消耗']=df['活动消耗'].round(2)

df['全天总消耗']=df['全天总消耗'].fillna(df['全天总消耗'].median())
df['全天总消耗']=df['全天总消耗'].round(2)
```

```python
#更新缺失基础代谢和活动消耗日期的全天总消耗列，计算方式为基础代谢+活动消耗
df.loc[df['日期']<'2025-11-01','全天总消耗']=round(df['基础代谢']+df['活动消耗'], 1)
```

```python
#终核数据是否有异常
df.describe()
```
